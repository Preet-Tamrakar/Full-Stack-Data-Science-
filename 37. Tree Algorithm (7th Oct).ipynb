# 🌿 1. Decision Tree Regression
🧠 Concept:

A Decision Tree splits the data into smaller groups based on conditions on features (like “Is Age < 30?”), aiming to minimize prediction error at each split.
Each leaf node gives a predicted value (usually the mean of target values in that region).

📘 How it works:

1) Start with the entire dataset.

Find the best feature and split point that minimizes error (like MSE).

Keep splitting until a stopping rule (like max depth) is met.

Each leaf node gives a prediction.
