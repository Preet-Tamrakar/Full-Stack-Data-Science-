# SUPPORT VECTOR REGRESSION (SVR)

Support Vector Regression (SVR) is a type of Support Vector Machine (SVM) used for regression problems — i.e., predicting continuous values (like price, salary, temperature, etc.).  
It tries to fit the best line (or hyperplane) within a margin of tolerance (ε) so that most data points lie within or near this margin.

== Support Vector Regression (Dataset is *Continuous*) - Regression SVR  
== Support Vector Machine (Datase is *Binary*) - Classification SVM
