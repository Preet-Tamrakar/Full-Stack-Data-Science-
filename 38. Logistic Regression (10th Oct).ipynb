# Logistic Regression 

+ Logistic Regression is a supervised machine learning algorithm used for classification problems.
+ It predicts the probability that a given input belongs to a particular category by applying the logistic (sigmoid) function to a linear combination of input features.

+ Textbook-style one-line definition: **Logistic Regression is a supervised machine learning algorithm used for classification tasks, which predicts the probability of a categorical dependent variable using the logistic (sigmoid) function.**

In simple terms:

It helps answer questions like —

+ “Will a customer buy the product — Yes or No?”
+ “Is this email spam or not?”

Key Points:

+ It outputs values between 0 and 1 (interpreted as probabilities).
+ A threshold (usually 0.5) is applied to classify into classes (e.g., 0 or 1).
+ Despite the name, it is used for classification, not regression.
+ Works well for binary and can be extended to multiclass classification (using One-vs-Rest).

Interview Question 

Que) Does Logistic Regression is Regression Algo or Classification Algo?

+ *Statement 1* --> If the dataset has **no outlier** it is called **Logistic Regression**.
+ *Statement 2* --> If the dataset has **outlier** logistic regression starts miss classification so now it called Logistic Classification. (We use Sigmoid to Correst the Curve)

Que) What is Regression ?

+ Regression Analysis is a predictive modelling techenique.
+ It estimates the relationship between a dependent (target) & an independent variable (predictor).

Que) Logistic Regression : What & Why?

+ Logistic Regression produces results in **binary format** which is used to predict the outcome of a catogrical dependent variable.
+ Outcome --> 0 or 1 | Yes or No | High or Low | True or False.
