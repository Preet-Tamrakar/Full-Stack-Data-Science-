# K Nearest Neighbour (KNN)

K-Nearest Neighbors (KNN) is a simple, intuitive, and non-parametric machine learning algorithm used for both classification and regression.

KNN Regression is a non-parametric, instance-based algorithm used to predict continuous values (like prices, temperatures, salaries, etc.).
It predicts the output of a new data point by averaging the target values of its K nearest neighbors in the training data.

KNN -- Distance matrix 2 type (distance 2 point)  
   + Euclidian Diatance : The shortest straight-line distance between two points in space.
   + Manhatten Distance : The sum of absolute differences between coordinates.
