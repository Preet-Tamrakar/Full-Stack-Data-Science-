**STOP WORDS --> Stop Words in NLP are Common, frequently occurring words that are filtered out before or during text processing because they add little to no meaning.**

```python
text
```

```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
stopwords = list(STOP_WORDS)
stopwords
```

```python
len(stopwords)
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
doc
```

**Problem Statement** -- From the document how to findout important sentence score?

Eg: IND vd AUS - INDIA own the Match
* Rohit & Virat (50 Over Played & Scored & both score 100)
* 8hrs Discussion.
* Rohit & Virat name repeates multiple times

Logic :
1. Document -tokens
2. token score - token frequency
3. weighted token frequency
4. document sentence
5. sentence == collection of tokens
6. token score = sentence score
7. score ascending -- top sentence scoring




```python
# lets get the tokens from text
tokens = [token.text for token in doc]
print(tokens)
# When we execute everything we created tokens from the text & not removed any of the stopwords
```

```python
doc
```

```python
# We have to calculate the frequency of each & every token, how many tokens is repetation in

word_frequencies = {}

for word in doc:
  if word.text.lower() not in stopwords:
    if word.text.lower() not in punctuation:
      if word.text not in word_frequencies.keys():
        word_frequencies[word.text] = 1
      else:
        word_frequencies[word.text] += 1
```

```python
word_frequencies
```

```python
len(word_frequencies)
```

```python
max_frequency = max(word_frequencies.values())
max_frequency
```

```python
# to get normalization/weighted frequencies you should devide all frequencies with 11

for word in word_frequencies.keys():
  word_frequencies[word] = word_frequencies[word]/max_frequency

# WEIGHTED SCORE / WEIGHTED FREQUENCY
```

```python
word_frequencies
```

```python
sentence_tokens = [sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
# We are going to calculate the sentence score , to calculate the sentence score

sentence_score = {}

for sent in sentence_tokens:
  for word in sent:
    if word.text.lower() in word_frequencies.keys():
      if sent not in sentence_score.keys():
        sentence_score[sent] = word_frequencies[word.text.lower()]
      else:
        sentence_score[sent] += word_frequencies[word.text.lower()]
```

```python
sentence_score
```

```python
len(sentence_score)
```

```python
# lets say our case study was 30% sentence with maximum score

from heapq import nlargest
```

```python
from numpy import select
select_length = int(len(sentence_tokens)*0.3)
select_length
```

```python
# We have to select maximum 40% sentences out of all sentence

summary = nlargest(select_length, sentence_score, key = sentence_score.get)
```

```python
summary
```

```python
# If I need to combine these top 3 sentence then

final_summary = [word.text for word in summary]
```
