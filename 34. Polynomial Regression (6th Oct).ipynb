# Polynomial Regression Formula: 

**y = β0​ + β1​x + β2​x^2 + β3​x^3 + ⋯ + βn​x^n + ϵ** 

+ y → Dependent variable (target/output)
+ x → Independent variable (input)
+ β0 (beta_0) → Intercept (constant term)
+ β1​+β2​+β3 (beta_1,beta_2,beta_3) → Coefficients for each power of ( x )
+ n → Degree of the polynomial
+ ϵ (epsilon) → Error term (difference between actual and predicted value)

### Hyperparameter Tuning
Hyperparameter tuning is the process of finding the best set of parameters for a machine learning model that control how the model learns.  
These are called hyperparameters because they are set before training, unlike model parameters (like weights) which are learned during training.
